```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
def plot_multiple_graphs(dataframe, columns, graphictype):
    for col in columns:
        plt.figure()
        plot_func = getattr(sn, graphictype)
        plot_func(data=dataframe, x='label', y=col)
        plt.xticks(rotation=90)
        plt.title(f'{col} do solo por cultura')
        plt.tight_layout()
        plt.show
```

```python
df = pd.read_csv('Crop_recommendation.csv')
columns = df.columns.tolist()
columns.remove('label')
df.groupby('label')[columns].mean()
```

```python
plt.Figure(figsize=(12,6))
crops = df.label.unique().tolist()
```

```python
plot_multiple_graphs(df, columns, 'boxplot')
```

```python
# Mapa de correlação entre features
sn.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```
